## Objectif
Ce notebook calcule l'**Indice de Qualité Économique (IQE)** pour chaque parcelle agricole du SCOT RPB.

## Pipeline
1. Chargement données RICA 2024 → PB/ha médian par code OTEX
2. Filtrage aberrants (50-500k €/ha) + scoring quintiles (1-5) → normalisation 0-1
3. Chargement RPG + clip sur SCOT RPB
4. Mapping RPG → RICA + attribution scores IQE
5. Export GeoPackage + CSV

**Référence** : CGAAER 2020 (IQE = 40% de l'IQAE)

````python
import pandas as pd
import geopandas as gpd
import numpy as np
import zipfile
import os
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

print('✅ Imports chargés')
````

````python
# ═══════════════════════════════════════════════════════════════
# CONFIGURATION - ADAPTER CES CHEMINS À VOTRE ENVIRONNEMENT
# ═══════════════════════════════════════════════════════════════

# Répertoire racine
REPERTOIRE = Path('.')  # Modifier si nécessaire

# Fichiers sources
FICHIER_RICA = REPERTOIRE / 'RICAMicroDonnees2024-v2'  # Dossier ou ZIP
FICHIER_RPG = REPERTOIRE / 'RPG_pac_RPB.gpkg'
FICHIER_SCOT = REPERTOIRE / 'TUP_SCoT_RPB.gpkg'

# Fichiers de sortie
SORTIE_GPKG = REPERTOIRE / 'RPG_IQE_SCOT_RPB.gpkg'
SORTIE_CSV = REPERTOIRE / 'IQE_scores_parcelles.csv'
SORTIE_REF = REPERTOIRE / 'Reference_scores_RICA.csv'

# Paramètres méthodologiques
SEUIL_MIN = 50          # PB/ha minimum (€)
SEUIL_MAX = 500000      # PB/ha maximum (€)
NB_QUINTILES = 5
CRS = 'EPSG:2154'       # Lambert-93

print('✅ Configuration chargée')
print(f'📁 Répertoire : {REPERTOIRE.resolve()}')
````

````python
# ═══════════════════════════════════════════════════════════════
# DICTIONNAIRE DE CORRESPONDANCE RPG → RICA (OTEX)
# ═══════════════════════════════════════════════════════════════

MAPPING_RPG_RICA = {
    # === CÉRÉALES (OTEX 15) ===
    'BTH': 15, 'BTP': 15, 'BDH': 15, 'BDP': 15,  # Blé
    'ORH': 15, 'ORP': 15,  # Orge
    'MIS': 15, 'MAS': 15,  # Maïs grain
    'TRN': 15, 'SGH': 15, 'AVH': 15, 'AVP': 15, 'SEG': 15,  # Autres céréales
    
    # === OLÉAGINEUX (OTEX 15) ===
    'CZH': 15, 'CZP': 15,  # Colza
    'SOJ': 15, 'TOU': 15,  # Soja, Tournesol
    
    # === PROTÉAGINEUX (OTEX 15) ===
    'PPR': 15, 'FEV': 15, 'LUP': 15, 'LEC': 15, 'PCH': 15,
    
    # === VITICULTURE (OTEX 35) ===
    'VRC': 35, 'VRT': 35,  # Vigne cuve/table
    
    # === FRUITS (OTEX 36) ===
    'PPE': 36, 'POI': 36, 'PCH': 36, 'ABR': 36, 'CER': 36,
    'PRU': 36, 'NOY': 36, 'CHA': 36, 'AMN': 36, 'OLI': 36,
    
    # === MARAÎCHAGE (OTEX 28) ===
    'MLN': 28, 'FRA': 28, 'LGF': 28, 'TOM': 28, 'CAR': 28,
    'SAL': 28, 'CHX': 28, 'PTG': 28, 'OIG': 28, 'AIL': 28,
    'LAV': 28, 'LAD': 28, 'PPM': 28,  # Lavande, lavandin, parfum
    
    # === PRAIRIES / FOURRAGES (OTEX 46) ===
    'PTR': 46, 'PPH': 46, 'PRL': 46,  # Prairies
    'SPH': 46, 'SPL': 46,  # Surfaces pastorales
    'MIF': 46, 'MFS': 46,  # Maïs fourrage
    'LUZ': 46, 'TRE': 46, 'SAF': 46, 'RGA': 46,  # Fourrages
    
    # === JACHÈRES / AUTRES ===
    'JAC': 0, 'SNE': 0, 'BFS': 0,  # Non valorisable
}

print(f'✅ Dictionnaire chargé : {len(MAPPING_RPG_RICA)} codes RPG')
````

---
## Étape 1 : Chargement et traitement des données RICA

````python
def charger_rica(chemin):
    '''
    Charge les microdonnées RICA depuis dossier, ZIP ou CSV.
    '''
    chemin = Path(chemin)
    
    # Cas ZIP
    if chemin.suffix == '.zip' and chemin.is_file():
        print(f'📦 Extraction ZIP : {chemin.name}')
        with zipfile.ZipFile(chemin, 'r') as z:
            csvs = [f for f in z.namelist() if f.endswith('.csv')]
            csv_file = max(csvs, key=lambda x: z.getinfo(x).file_size)
            with z.open(csv_file) as f:
                df = pd.read_csv(f, sep=';', low_memory=False)
    
    # Cas dossier
    elif chemin.is_dir():
        print(f'📂 Lecture dossier : {chemin.name}')
        csvs = list(chemin.glob('**/*.csv'))
        csv_file = max(csvs, key=lambda x: x.stat().st_size)
        print(f'   → Fichier : {csv_file.name}')
        df = pd.read_csv(csv_file, sep=';', low_memory=False)
    
    # Cas CSV direct
    elif chemin.suffix == '.csv':
        df = pd.read_csv(chemin, sep=';', low_memory=False)
    
    else:
        raise FileNotFoundError(f'Données RICA introuvables : {chemin}')
    
    print(f'   ✅ {len(df):,} lignes, {len(df.columns)} colonnes')
    return df

# Chargement
df_rica = charger_rica(FICHIER_RICA)
print('\n📋 Aperçu colonnes :', df_rica.columns.tolist()[:15])
````

````python
def identifier_colonnes(df):
    '''
    Identifie automatiquement les colonnes OTEX, PB, SAU dans les données RICA.
    '''
    cols = {}
    col_list = df.columns.tolist()
    
    # OTEX
    for p in ['OTEX', 'OTEX8', 'TYPE_EXPLOIT']:
        matches = [c for c in col_list if p.upper() in c.upper()]
        if matches:
            cols['otex'] = matches[0]
            break
    
    # Produit Brut
    for p in ['PRODBRUT', 'SE131', 'PB_TOTAL']:
        matches = [c for c in col_list if p.upper() in c.upper()]
        if matches:
            cols['pb'] = matches[0]
            break
    
    # SAU
    for p in ['SAUTOT', 'SE025', 'SAU']:
        matches = [c for c in col_list if p.upper() in c.upper()]
        if matches:
            cols['sau'] = matches[0]
            break
    
    print('🔍 Colonnes identifiées :')
    for k, v in cols.items():
        print(f'   • {k} → {v}')
    
    return cols

cols_rica = identifier_colonnes(df_rica)
````

````python
def calculer_pb_ha(df, cols):
    '''
    Calcule le PB/ha médian par code OTEX.
    '''
    df = df.copy()
    
    # Conversion numérique
    df[cols['pb']] = pd.to_numeric(df[cols['pb']], errors='coerce')
    df[cols['sau']] = pd.to_numeric(df[cols['sau']], errors='coerce')
    
    # Calcul PB/ha
    df['PB_HA'] = df[cols['pb']] / df[cols['sau']]
    
    # Filtrage aberrants
    n0 = len(df)
    df = df[(df['PB_HA'] >= SEUIL_MIN) & (df['PB_HA'] <= SEUIL_MAX) & df['PB_HA'].notna()]
    print(f'🔧 Filtrage : {n0:,} → {len(df):,} ({n0-len(df):,} exclus)')
    
    # Agrégation par OTEX
    agg = df.groupby(cols['otex']).agg(
        PB_HA_MEDIAN=('PB_HA', 'median'),
        PB_HA_MEAN=('PB_HA', 'mean'),
        N_EXPLOIT=('PB_HA', 'count')
    ).reset_index()
    agg = agg.rename(columns={cols['otex']: 'CODE_RICA'})
    
    print(f'✅ {len(agg)} codes RICA traités')
    return agg

df_pb = calculer_pb_ha(df_rica, cols_rica)
df_pb.head(10)
````

---
## Étape 2 : Scoring en quintiles et normalisation

````python
def calculer_score_iqe(df):
    '''
    Score en quintiles (1-5) puis normalisation 0-1.
    '''
    df = df.copy()
    
    # Quintiles
    df['SCORE_QUINTILE'] = pd.qcut(
        df['PB_HA_MEDIAN'],
        q=NB_QUINTILES,
        labels=range(1, NB_QUINTILES + 1),
        duplicates='drop'
    ).astype(int)
    
    # Normalisation 0-1
    df['IQE'] = (df['SCORE_QUINTILE'] - 1) / (NB_QUINTILES - 1)
    
    print('📊 Distribution scores :')
    print(df.groupby('SCORE_QUINTILE').agg(
        N=('CODE_RICA', 'count'),
        PB_min=('PB_HA_MEDIAN', 'min'),
        PB_max=('PB_HA_MEDIAN', 'max')
    ).round(0))
    
    return df

df_scores = calculer_score_iqe(df_pb)
print('\n✅ Table de référence RICA :')
df_scores
````

---
## Étape 3 : Chargement des données spatiales

````python
# Chargement du périmètre SCOT
print('🗺️ Chargement SCOT RPB...')
gdf_scot = gpd.read_file(FICHIER_SCOT)
gdf_scot = gdf_scot.to_crs(CRS)
print(f'   ✅ {len(gdf_scot)} polygones, CRS = {gdf_scot.crs}')

# Union pour avoir un seul polygone
scot_union = gdf_scot.union_all()
print(f'   Surface totale : {scot_union.area/1e6:.1f} km²')
````

````python
# Chargement du RPG
print('🌾 Chargement RPG...')
gdf_rpg = gpd.read_file(FICHIER_RPG)
gdf_rpg = gdf_rpg.to_crs(CRS)
print(f'   ✅ {len(gdf_rpg):,} parcelles, CRS = {gdf_rpg.crs}')
print(f'   Colonnes : {gdf_rpg.columns.tolist()}')
````

````python
# Identifier la colonne code culture
def trouver_col_culture(gdf):
    for c in ['CODE_CULTU', 'CODE_GROUP', 'CODE_CULTURE', 'CULT_MAJ', 'CODE']:
        matches = [col for col in gdf.columns if c.upper() in col.upper()]
        if matches:
            return matches[0]
    return None

COL_CULTURE = trouver_col_culture(gdf_rpg)
print(f'📌 Colonne culture : {COL_CULTURE}')

# Aperçu des codes présents
print('\n🌿 Top 15 codes culture :')
print(gdf_rpg[COL_CULTURE].value_counts().head(15))
````

---
## Étape 4 : Clip spatial sur le périmètre SCOT

````python
# Clip des parcelles sur le SCOT
print('✂️ Clip spatial RPG sur SCOT...')
gdf_rpg_scot = gpd.clip(gdf_rpg, gdf_scot)
print(f'   ✅ {len(gdf_rpg_scot):,} parcelles dans le périmètre SCOT')

# Supprimer géométries vides
gdf_rpg_scot = gdf_rpg_scot[~gdf_rpg_scot.geometry.is_empty]
print(f'   Après nettoyage : {len(gdf_rpg_scot):,} parcelles')
````

---
## Étape 5 : Attribution des scores IQE aux parcelles

````python
def attribuer_scores(gdf, df_scores, col_culture, mapping):
    '''
    Attribue le score IQE à chaque parcelle via le mapping RPG→RICA.
    '''
    gdf = gdf.copy()
    
    # Mapping RPG → RICA
    gdf['CODE_RICA'] = gdf[col_culture].map(mapping)
    
    # Statistiques mapping
    n_mapped = gdf['CODE_RICA'].notna().sum()
    n_total = len(gdf)
    print(f'🔗 Mapping RPG→RICA : {n_mapped:,}/{n_total:,} ({100*n_mapped/n_total:.1f}%)')
    
    # Codes non mappés
    non_mappes = gdf[gdf['CODE_RICA'].isna()][col_culture].unique()
    if len(non_mappes) > 0:
        print(f'   ⚠️ Codes non mappés : {list(non_mappes)[:10]}')
    
    # Jointure avec scores
    df_scores_clean = df_scores[['CODE_RICA', 'PB_HA_MEDIAN', 'SCORE_QUINTILE', 'IQE']].copy()
    gdf = gdf.merge(df_scores_clean, on='CODE_RICA', how='left')
    
    # Statistiques finales
    n_scored = gdf['IQE'].notna().sum()
    print(f'✅ Parcelles avec IQE : {n_scored:,}/{n_total:,} ({100*n_scored/n_total:.1f}%)')
    
    return gdf

gdf_final = attribuer_scores(gdf_rpg_scot, df_scores, COL_CULTURE, MAPPING_RPG_RICA)
````

````python
# Statistiques finales
print('📊 STATISTIQUES FINALES')
print('=' * 50)

# Distribution IQE
print('\n🎯 Distribution des scores IQE :')
iqe_stats = gdf_final.groupby('SCORE_QUINTILE').agg(
    N_parcelles=('IQE', 'count'),
    Surface_ha=('geometry', lambda x: x.area.sum() / 10000)
).round(1)
print(iqe_stats)

# Surface totale
surf_totale = gdf_final.geometry.area.sum() / 10000
surf_scoree = gdf_final[gdf_final['IQE'].notna()].geometry.area.sum() / 10000
print(f'\n📐 Surface totale : {surf_totale:,.0f} ha')
print(f'   Surface avec IQE : {surf_scoree:,.0f} ha ({100*surf_scoree/surf_totale:.1f}%)')

# IQE moyen
iqe_moyen = gdf_final['IQE'].mean()
print(f'\n📈 IQE moyen : {iqe_moyen:.3f}')
````

---
## Étape 6 : Export des résultats

````python
# Export GeoPackage
print('💾 Export des résultats...')

# GeoPackage avec géométries
gdf_final.to_file(SORTIE_GPKG, driver='GPKG')
print(f'   ✅ GeoPackage : {SORTIE_GPKG}')

# CSV sans géométries
cols_export = [c for c in gdf_final.columns if c != 'geometry']
gdf_final[cols_export].to_csv(SORTIE_CSV, index=False, encoding='utf-8')
print(f'   ✅ CSV parcelles : {SORTIE_CSV}')

# Table de référence RICA
df_scores.to_csv(SORTIE_REF, index=False, encoding='utf-8')
print(f'   ✅ Référence RICA : {SORTIE_REF}')

print('\n🎉 TRAITEMENT TERMINÉ')
````

## Résumé méthodologique

### Sources de données
| Donnée | Source | Usage |
|--------|--------|-------|
| RICA 2024 | Agreste | PB/ha par type d'exploitation |
| RPG | IGN/ASP | Géolocalisation parcelles agricoles |
| SCOT RPB | DDT | Périmètre d'étude |

### Pipeline de traitement
1. **Extraction RICA** : PB/ha médian par code OTEX
2. **Filtrage** : 50 ≤ PB/ha ≤ 500 000 €/ha
3. **Scoring** : Quintiles 1-5 → normalisation 0-1
4. **Mapping** : Code RPG → Code RICA (OTEX)
5. **Jointure spatiale** : Clip RPG sur SCOT + attribution scores

### Formule IQE
```
IQE = (Quintile - 1) / 4
```
Où Quintile ∈ {1, 2, 3, 4, 5} basé sur le PB/ha médian du type de culture.

### Référence
CGAAER (2020) - *Méthodologie de calcul de l'IQAE* : **IQE = 40% de l'IQAE final**
