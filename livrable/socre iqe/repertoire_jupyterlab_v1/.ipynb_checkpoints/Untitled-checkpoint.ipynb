```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration pour de beaux graphiques de présentation
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("talk") # Taille de police adaptée pour les slides

# 1. CHARGEMENT DES DONNÉES
print("Chargement du RPG...")
# On charge le fichier spatial
rpg = gpd.read_file('RPG_pac_RPB.gpkg')

# Vérification des colonnes disponibles (à adapter selon les vrais noms de tes colonnes)
print("Colonnes disponibles :", rpg.columns.tolist())
```

```python
import geopandas as gpd

# 1. Chargement

communes = gpd.read_file('ADE_4-0_GPKG_LAMB93_FXX-ED2026-02-16.gpkg', layer='commune')

# 2. Jointure spatiale (en s'assurant que tout est en Lambert 93 - EPSG:2154)
rpg = rpg.to_crs(epsg=2154)
communes = communes.to_crs(epsg=2154)

```

```python
communes.columns
```

```python
# On sélectionne les bonnes colonnes de ton fichier Admin Express et on les renomme
communes_light = communes[['code_insee', 'nom_officiel', 'code_insee_de_la_region', 'geometry']].rename(
    columns={
        'code_insee': 'INSEE_CORRIGE', 
        'nom_officiel': 'NOM_COMMUNE_CORRIGE',
        'code_insee_de_la_region': 'CODE_REGION'
    }
)

# On effectue la jointure spatiale
rpg_corrige = gpd.sjoin(rpg, communes_light, how='left', predicate='intersects')

# On supprime les doublons (parcelles à la frontière stricte de deux communes)
rpg_corrige = rpg_corrige[~rpg_corrige.index.duplicated(keep='first')]

# --- Pour vérifier que ça a bien marché ---
print("Extrait après jointure :")
display(rpg_corrige[['INSEE_CORRIGE', 'NOM_COMMUNE_CORRIGE', 'CODE_REGION']].head())
```

```python
rpg = rpg_corrige.copy()
```

```python
# 1. Vérification et Reprojection en Lambert 93 (EPSG:2154)
# C'est la projection officielle française en mètres. Indispensable pour un calcul de surface juste.
if rpg.crs is None or rpg.crs.to_epsg() != 2154:
    print(f"Système de coordonnées actuel : {rpg.crs}. Reprojection en Lambert 93 (EPSG:2154)...")
    rpg = rpg.to_crs(epsg=2154)

# 2. RECALCUL DE LA SURFACE GÉOMÉTRIQUE
print("Recalcul manuel des surfaces géométriques des parcelles...")
# rpg.geometry.area donne la surface en m² (car le Lambert 93 est en mètres)
# On divise par 10 000 pour avoir des hectares
rpg['surf_ha_calculee'] = rpg.geometry.area / 10000

# Petit test pour voir la différence avec l'ancienne colonne
diff_moyenne = (rpg['surf_ha_calculee'] - pd.to_numeric(rpg['surf_ha'], errors='coerce').fillna(0)).mean()
print(f"La différence moyenne entre la surface déclarée et la géométrie réelle est de {diff_moyenne:.4f} hectares par parcelle.")

# 3. Création des Départements (à partir du code INSEE)
rpg['code_dep'] = rpg['insee'].astype(str).str[:2]
dict_dep = {'07': 'Ardèche', '26': 'Drôme', '84': 'Vaucluse'}
rpg['nom_dep'] = rpg['code_dep'].map(dict_dep).fillna('Autre')

print("\nAperçu des nouvelles surfaces calculées :")
print(rpg[['insee', 'nom', 'code_culture', 'surf_ha_calculee']].head())
```

```python
import folium
import geopandas as gpd

print("1. Filtrage sur une zone test (en Lambert 93 pour le calcul du centre)...")
commune_test = 'Montélimar' 

if commune_test not in rpg_corrige['NOM_COMMUNE_CORRIGE'].values:
    commune_test = rpg_corrige['NOM_COMMUNE_CORRIGE'].dropna().iloc[0]
    print(f"Commune choisie par defaut : {commune_test}")

rpg_test = rpg_corrige[rpg_corrige['NOM_COMMUNE_CORRIGE'] == commune_test].copy()

# Calcul du centre
centre_x = rpg_test.geometry.centroid.x.mean()
centre_y = rpg_test.geometry.centroid.y.mean()

# Conversion en GPS
centre_point = gpd.GeoSeries(gpd.points_from_xy([centre_x], [centre_y]), crs="EPSG:2154").to_crs(epsg=4326)
centre_lat = centre_point.y.iloc[0]
centre_lon = centre_point.x.iloc[0]

print(f"Centre : {centre_lat:.4f}, {centre_lon:.4f}")

print("2. Reprojection des donnees en coordonnees GPS (EPSG:4326)...")
rpg_gps = rpg_test.to_crs(epsg=4326)

print("3. Creation de la carte interactive Folium...")

# Limites France metropolitaine pour eviter de zoomer ailleurs
limites_france = [[41.3, -5.2], [51.1, 9.6]]

ma_carte = folium.Map(
    location=[centre_lat, centre_lon], 
    zoom_start=13,
    min_zoom=5,
    max_bounds=True,
    bounds=limites_france
)

# Fond IGN France (Plan IGN)
folium.TileLayer(
    tiles='https://wxs.ign.fr/choisirgeoportail/geoportail/wmts?SERVICE=WMTS&REQUEST=GetTile&VERSION=1.0.0&LAYER=GEOGRAPHICALGRIDSYSTEMS.PLANIGNV2&STYLE=normal&FORMAT=image/png&TILEMATRIXSET=PM&TILEMATRIX={z}&TILEROW={y}&TILECOL={x}',
    attr='IGN France',
    name='Plan IGN',
    overlay=False,
    control=True
).add_to(ma_carte)

# Fond IGN Orthophoto (satellite France)
folium.TileLayer(
    tiles='https://wxs.ign.fr/choisirgeoportail/geoportail/wmts?SERVICE=WMTS&REQUEST=GetTile&VERSION=1.0.0&LAYER=ORTHOIMAGERY.ORTHOPHOTOS&STYLE=normal&FORMAT=image/jpeg&TILEMATRIXSET=PM&TILEMATRIX={z}&TILEROW={y}&TILECOL={x}',
    attr='IGN France',
    name='Photo aerienne IGN',
    overlay=False,
    control=True
).add_to(ma_carte)

# Fond OpenStreetMap France
folium.TileLayer(
    tiles='https://{s}.tile.openstreetmap.fr/osmfr/{z}/{x}/{y}.png',
    attr='OpenStreetMap France',
    name='OSM France',
    overlay=False,
    control=True
).add_to(ma_carte)

print("4. Ajout des parcelles agricoles...")
folium.GeoJson(
    rpg_gps,
    name='Parcelles RPG',
    style_function=lambda feature: {
        'fillColor': '#ffeb3b',   
        'color': '#f44336',       
        'weight': 2,              
        'fillOpacity': 0.4,       
    },
    tooltip=folium.GeoJsonTooltip(
        fields=['NOM_COMMUNE_CORRIGE', 'code_culture', 'surf_ha_calculee'],
        aliases=['Commune:', 'Code Culture PAC:', 'Surface (ha):'],
        style="background-color: white; color: #333333; font-family: arial; font-size: 12px; padding: 10px;"
    )
).add_to(ma_carte)

folium.LayerControl().add_to(ma_carte)
print("Carte generee !")
ma_carte
```

```python
import pandas as pd

print("1. Chargement du referentiel des cultures 2023...")
# Ajout de l'encodage latin-1 pour lire correctement les accents francais
ref_df = pd.read_csv('REF_CULTURES_GROUPES_CULTURES_2023.csv', sep=';', encoding='latin-1')

# On s'assure qu'il n'y a pas de doublons dans le referentiel pour eviter de dupliquer les parcelles
ref_df = ref_df.drop_duplicates(subset=['CODE_CULTURE'])

print("2. Preparation et Jointure avec le RPG...")
# On force le format texte et on enleve les espaces pour assurer une jointure parfaite
rpg['code_culture'] = rpg['code_culture'].astype(str).str.strip()
ref_df['CODE_CULTURE'] = ref_df['CODE_CULTURE'].astype(str).str.strip()

# On realise la jointure (Left Join)
rpg = rpg.merge(
    ref_df[['CODE_CULTURE', 'LIBELLE_CULTURE', 'LIBELLE_GROUPE_CULTURE']], 
    left_on='code_culture', 
    right_on='CODE_CULTURE', 
    how='left'
)

print("\n3. Calcul des taux de mapping...")
# Calcul des statistiques
total_parcelles = len(rpg)
match_culture = rpg['LIBELLE_CULTURE'].notna().sum()
match_groupe = rpg['LIBELLE_GROUPE_CULTURE'].notna().sum()

taux_culture = (match_culture / total_parcelles) * 100
taux_groupe = (match_groupe / total_parcelles) * 100

print("-" * 50)
print("--- RESULTATS DU MAPPING (QUALITE DES DONNEES) ---")
print("-" * 50)
print(f"Nombre total de parcelles dans le RPG : {total_parcelles:,.0f}".replace(',', ' '))
print(f"Taux de libelles Cultures traduits    : {taux_culture:.2f} %")
print(f"Taux de libelles Groupes traduits     : {taux_groupe:.2f} %")
print("-" * 50)

# 4. Investigation des erreurs (les codes qui n'ont pas matche)
if taux_culture < 100:
    print("\nATTENTION : Certains codes du RPG n'ont pas ete trouves dans le referentiel 2023.")
    print("Voici le Top 10 des codes orphelins (avec leur nombre de parcelles) :")
    
    # On isole les lignes ou le libelle est manquant
    orphelins = rpg[rpg['LIBELLE_CULTURE'].isna()]
    
    # On compte les codes qui posent probleme
    valeurs_manquantes = orphelins['code_culture'].value_counts()
    print(valeurs_manquantes.head(10))
    
    # Impact en surface
    surface_orpheline = orphelins['surf_ha_calculee'].sum()
    print(f"\n=> Cela represente une surface agricole 'inconnue' de {surface_orpheline:.1f} hectares.")
else:
    print("\nMapping parfait ! 100% des parcelles ont desormais un nom de culture et un groupe en clair.")

# Apercu du resultat final
print("\nApercu de la base PAC enrichie :")
display(rpg[['insee', 'code_culture', 'LIBELLE_CULTURE', 'LIBELLE_GROUPE_CULTURE', 'surf_ha_calculee']].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Calcul de la répartition des cultures en cours...")

# 1. Calcul de la surface agricole totale du SCoT
surface_totale_agricole = rpg['surf_ha_calculee'].sum()

# 2. Regroupement par 'code_group' et calcul des surfaces
df_cultures = rpg.groupby('LIBELLE_GROUPE_CULTURE').agg(
    surface_totale_ha=('surf_ha_calculee', 'sum'),
    nb_parcelles=('surf_ha_calculee', 'count')
).reset_index()

# 3. Calcul du poids (en %) de chaque culture sur le territoire
df_cultures['part_pourcentage'] = (df_cultures['surface_totale_ha'] / surface_totale_agricole) * 100

# 4. Tri du plus grand au plus petit
df_cultures = df_cultures.sort_values(by='surface_totale_ha', ascending=False)

# Affichage du tableau propre (DataFrame)
print(f"\nSurface totale analysée : {surface_totale_agricole:,.0f} hectares")
print("\n--- TABLEAU COMPLET DES CULTURES (Aperçu) ---")
# 'display' affiche un beau tableau dans Jupyter, sinon on utilise print()
display(df_cultures.head(15)) 

```

```python
# =============================================================================
# JOINTURE PBS ET CALCUL IQE NORMALISE (0-1)
# =============================================================================

# Chargement du dictionnaire deja construit
dico = pd.read_csv('DICTIONNAIRE_RPG_CPS_SCOT_RPB.csv', sep=';')

# Correspondance departement vers region
DEP_REGION = {'07': 'RA', '26': 'RA', '84': 'PACA'}

# Creation des dictionnaires de lookup
dico_pbs_ra = dict(zip(dico['CODE_RPG'], dico['PBS_RHONE_ALPES']))
dico_pbs_paca = dict(zip(dico['CODE_RPG'], dico['PBS_PACA']))

# Ajout de la PBS selon le departement
def get_pbs(row):
    code = row['code_culture']
    code_dep = str(row['code_dep'])
    region = DEP_REGION.get(code_dep, 'RA')
    if region == 'RA':
        return dico_pbs_ra.get(code, 0.0)
    else:
        return dico_pbs_paca.get(code, 0.0)

rpg['PBS_HA'] = rpg.apply(get_pbs, axis=1)
rpg['PBS_TOTALE'] = rpg['PBS_HA'] * rpg['surf_ha_calculee']

# Normalisation IQE de 0 a 1 (min-max)
pbs_min = rpg['PBS_HA'].min()
pbs_max = rpg['PBS_HA'].max()
rpg['IQE_NORMALISE'] = (rpg['PBS_HA'] - pbs_min) / (pbs_max - pbs_min)

# Score IQE classique (1-5) selon CGAAER
def score_iqe(pbs):
    if pbs >= 15000: return 5
    elif pbs >= 8000: return 4
    elif pbs >= 4000: return 3
    elif pbs >= 2000: return 2
    else: return 1

rpg['SCORE_IQE'] = rpg['PBS_HA'].apply(score_iqe)

print("Jointure PBS terminee")
print(f"Parcelles : {len(rpg)}")
print(f"PBS min : {pbs_min:.2f} euros/ha")
print(f"PBS max : {pbs_max:.2f} euros/ha")
print(f"PBS moyenne : {rpg['PBS_HA'].mean():.2f} euros/ha")
```

```python
# =============================================================================
# ANALYSE COMPLETE IQE - SCOT RPB
# =============================================================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.dpi'] = 150
plt.rcParams['font.size'] = 10

# =============================================================================
# 1. STATISTIQUES DESCRIPTIVES GLOBALES
# =============================================================================

print("=" * 70)
print("ANALYSE IQE - SCOT RHONE PROVENCE BARONNIES")
print("=" * 70)

print("\n1. RESUME GLOBAL")
print("-" * 50)
print(f"Nombre total de parcelles : {len(rpg):,}")
print(f"Surface agricole totale : {rpg['surf_ha_calculee'].sum():,.1f} ha")
print(f"Surface moyenne par parcelle : {rpg['surf_ha_calculee'].mean():.2f} ha")
print(f"Surface mediane par parcelle : {rpg['surf_ha_calculee'].median():.2f} ha")

print("\n2. STATISTIQUES PBS")
print("-" * 50)
print(f"PBS minimum : {rpg['PBS_HA'].min():,.2f} euros/ha")
print(f"PBS maximum : {rpg['PBS_HA'].max():,.2f} euros/ha")
print(f"PBS moyenne : {rpg['PBS_HA'].mean():,.2f} euros/ha")
print(f"PBS mediane : {rpg['PBS_HA'].median():,.2f} euros/ha")
print(f"PBS ecart-type : {rpg['PBS_HA'].std():,.2f} euros/ha")
print(f"PBS totale territoire : {rpg['PBS_TOTALE'].sum()/1e6:,.2f} millions euros")

print("\n3. STATISTIQUES IQE NORMALISE (0-1)")
print("-" * 50)
print(f"IQE moyen : {rpg['IQE_NORMALISE'].mean():.4f}")
print(f"IQE median : {rpg['IQE_NORMALISE'].median():.4f}")
print(f"IQE ecart-type : {rpg['IQE_NORMALISE'].std():.4f}")

print("\n4. DISTRIBUTION DES SCORES IQE")
print("-" * 50)
print(f"{'Score':<10} {'Parcelles':>12} {'%':>8} {'Surface (ha)':>15} {'% surf':>8}")
print("-" * 50)
for score in [5, 4, 3, 2, 1]:
    n = len(rpg[rpg['SCORE_IQE'] == score])
    pct = n / len(rpg) * 100
    surf = rpg[rpg['SCORE_IQE'] == score]['surf_ha_calculee'].sum()
    pct_surf = surf / rpg['surf_ha_calculee'].sum() * 100
    print(f"Score {score:<4} {n:>12,} {pct:>7.1f}% {surf:>14,.1f} {pct_surf:>7.1f}%")

# =============================================================================
# 2. ANALYSE PAR DEPARTEMENT
# =============================================================================

print("\n" + "=" * 70)
print("ANALYSE PAR DEPARTEMENT")
print("=" * 70)

stats_dep = rpg.groupby('nom_dep').agg({
    'surf_ha_calculee': ['sum', 'mean', 'count'],
    'PBS_HA': ['mean', 'median', 'std'],
    'PBS_TOTALE': 'sum',
    'IQE_NORMALISE': ['mean', 'median']
}).round(2)

stats_dep.columns = ['Surface totale (ha)', 'Surface moy (ha)', 'Nb parcelles',
                     'PBS moy', 'PBS med', 'PBS std', 
                     'PBS totale', 'IQE moy', 'IQE med']
stats_dep['PBS totale (M)'] = (stats_dep['PBS totale'] / 1e6).round(2)
stats_dep = stats_dep.drop('PBS totale', axis=1)
print(stats_dep)

# Tableau croise scores x departements
print("\n5. REPARTITION SCORES IQE PAR DEPARTEMENT (nb parcelles)")
print("-" * 50)
pivot_count = pd.crosstab(rpg['SCORE_IQE'], rpg['nom_dep'], margins=True)
print(pivot_count)

print("\n6. REPARTITION SCORES IQE PAR DEPARTEMENT (surface ha)")
print("-" * 50)
pivot_surf = rpg.pivot_table(values='surf_ha_calculee', index='SCORE_IQE', 
                              columns='nom_dep', aggfunc='sum', margins=True).round(1)
print(pivot_surf)

# =============================================================================
# 3. ANALYSE PAR CULTURE
# =============================================================================

print("\n" + "=" * 70)
print("ANALYSE PAR TYPE DE CULTURE")
print("=" * 70)

stats_culture = rpg.groupby('code_culture').agg({
    'surf_ha_calculee': ['sum', 'count'],
    'PBS_HA': 'mean',
    'SCORE_IQE': 'mean'
}).round(2)
stats_culture.columns = ['Surface (ha)', 'Nb parcelles', 'PBS moy', 'Score moy']
stats_culture = stats_culture.sort_values('Surface (ha)', ascending=False)

print("\n7. TOP 15 CULTURES PAR SURFACE")
print("-" * 50)
top_surf = stats_culture.head(15)
for code, row in top_surf.iterrows():
    lib = rpg[rpg['code_culture'] == code]['LIBELLE_CULTURE'].iloc[0] if 'LIBELLE_CULTURE' in rpg.columns else code
    lib = str(lib)[:35]
    print(f"{code:5} {lib:<35} {row['Surface (ha)']:>10,.1f} ha  PBS: {row['PBS moy']:>10,.0f}")

print("\n8. TOP 15 CULTURES PAR PBS MOYENNE")
print("-" * 50)
top_pbs = stats_culture.sort_values('PBS moy', ascending=False).head(15)
for code, row in top_pbs.iterrows():
    lib = rpg[rpg['code_culture'] == code]['LIBELLE_CULTURE'].iloc[0] if 'LIBELLE_CULTURE' in rpg.columns else code
    lib = str(lib)[:35]
    print(f"{code:5} {lib:<35} PBS: {row['PBS moy']:>10,.0f}  {row['Surface (ha)']:>8,.1f} ha")

# =============================================================================
# 4. ANALYSE PAR GROUPE DE CULTURE
# =============================================================================

if 'LIBELLE_GROUPE_CULTURE' in rpg.columns:
    print("\n" + "=" * 70)
    print("ANALYSE PAR GROUPE DE CULTURE")
    print("=" * 70)
    
    stats_groupe = rpg.groupby('LIBELLE_GROUPE_CULTURE').agg({
        'surf_ha_calculee': ['sum', 'count'],
        'PBS_HA': 'mean',
        'PBS_TOTALE': 'sum',
        'IQE_NORMALISE': 'mean'
    }).round(2)
    stats_groupe.columns = ['Surface (ha)', 'Nb parcelles', 'PBS moy', 'PBS totale', 'IQE moy']
    stats_groupe = stats_groupe.sort_values('Surface (ha)', ascending=False)
    
    print("\n9. GROUPES DE CULTURE PAR SURFACE")
    print("-" * 60)
    for groupe, row in stats_groupe.head(15).iterrows():
        groupe_str = str(groupe)[:40]
        print(f"{groupe_str:<40} {row['Surface (ha)']:>10,.1f} ha  PBS: {row['PBS moy']:>8,.0f}  IQE: {row['IQE moy']:.3f}")

# =============================================================================
# 5. INDICATEURS CLES POUR PRESENTATION
# =============================================================================

print("\n" + "=" * 70)
print("INDICATEURS CLES POUR PRESENTATION")
print("=" * 70)

# Cultures a forte valeur (score 5)
score5 = rpg[rpg['SCORE_IQE'] == 5]
print(f"\nCULTURES A TRES HAUTE VALEUR (Score 5 >= 15000 euros/ha)")
print(f"  - Nombre de parcelles : {len(score5):,} ({len(score5)/len(rpg)*100:.1f}%)")
print(f"  - Surface : {score5['surf_ha_calculee'].sum():,.1f} ha ({score5['surf_ha_calculee'].sum()/rpg['surf_ha_calculee'].sum()*100:.1f}%)")
print(f"  - PBS totale : {score5['PBS_TOTALE'].sum()/1e6:.2f} M euros ({score5['PBS_TOTALE'].sum()/rpg['PBS_TOTALE'].sum()*100:.1f}%)")

# Cultures a faible valeur (score 1)
score1 = rpg[rpg['SCORE_IQE'] == 1]
print(f"\nCULTURES A FAIBLE VALEUR (Score 1 < 2000 euros/ha)")
print(f"  - Nombre de parcelles : {len(score1):,} ({len(score1)/len(rpg)*100:.1f}%)")
print(f"  - Surface : {score1['surf_ha_calculee'].sum():,.1f} ha ({score1['surf_ha_calculee'].sum()/rpg['surf_ha_calculee'].sum()*100:.1f}%)")
print(f"  - PBS totale : {score1['PBS_TOTALE'].sum()/1e6:.2f} M euros ({score1['PBS_TOTALE'].sum()/rpg['PBS_TOTALE'].sum()*100:.1f}%)")

# Comparaison departements
print(f"\nCOMPARAISON DEPARTEMENTS")
for dep in rpg['nom_dep'].unique():
    dep_data = rpg[rpg['nom_dep'] == dep]
    print(f"  {dep}:")
    print(f"    - PBS moyenne : {dep_data['PBS_HA'].mean():,.0f} euros/ha")
    print(f"    - IQE moyen : {dep_data['IQE_NORMALISE'].mean():.3f}")
    print(f"    - % parcelles score 5 : {len(dep_data[dep_data['SCORE_IQE']==5])/len(dep_data)*100:.1f}%")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Calcul de l'Indice de Compacité (Vulnérabilité morphologique)...")

# 1. Calcul du périmètre géométrique (en mètres)
rpg['perimetre_m'] = rpg.geometry.length

# 2. Calcul de la surface en mètres carrés (déjà utilisée tout à l'heure)
rpg['surface_m2'] = rpg.geometry.area

# 3. Formule de l'Indice de Compacité (Périmètre / (2 * sqrt(pi * Surface)))
# Plus l'indice est éloigné de 1, plus la parcelle est biscornue et vulnérable.
rpg['indice_vulnerabilite'] = rpg['perimetre_m'] / (2 * np.sqrt(np.pi * rpg['surface_m2']))

# On retire les éventuelles parcelles microscopiques (artefacts SIG) qui faussent les stats
rpg_propre = rpg[rpg['surface_m2'] > 100].copy()

# 4. Classification des parcelles
def classer_vulnerabilite(indice):
    if indice < 1.15: return '1. Très Compacte (Sûre)'
    elif indice < 1.30: return '2. Normale'
    elif indice < 1.60: return '3. Irrégulière (Vulnérable)'
    else: return '4. Très Éclatée (Très Vulnérable)'

rpg_propre['classe_vuln'] = rpg_propre['indice_vulnerabilite'].apply(classer_vulnerabilite)

print("\nRépartition de la vulnérabilité des parcelles du SCoT :")
print(rpg_propre['classe_vuln'].value_counts())

# ==========================================
# GRAPHIQUE 1 : La part des parcelles vulnérables
# ==========================================
plt.figure(figsize=(8, 8))
colors_vuln = ['#2ca02c', '#bcbd22', '#ff7f0e', '#d62728'] # Du vert (sûr) au rouge (vulnérable)

repartition = rpg_propre['classe_vuln'].value_counts().sort_index()
plt.pie(repartition, labels=repartition.index, autopct='%1.1f%%', 
        colors=colors_vuln, startangle=90, explode=[0, 0, 0.05, 0.1])
plt.title("Vulnérabilité Morphologique des Parcelles Agricoles", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('05_Camembert_Vulnerabilite.png', dpi=300)
plt.show()

```
